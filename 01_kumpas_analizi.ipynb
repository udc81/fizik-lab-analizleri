```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
olcum_no = np.arange(1, 11)
kumpas_degerleri = np.array([2.008, 2.010, 2.30, 2.10, 2.03, 2.02, 2.04, 2.01, 2.06, 2.02])
```

```python
# Kumpas Grafigi
ortalama = np.mean(kumpas_degerleri)
plt.plot(olcum_no, kumpas_degerleri, 'o', color='blue', label='Kumpas Ölçümleri')
plt.axhline(ortalama, color='red', linestyle='--', label=f'Ortalama: {ortalama:.3f} cm')
plt.xlabel('Ölçüm Sırası')
plt.ylabel('Uzunluk (cm)')
plt.title('Kumpas Ölçümleri ve Ortalama Değer')
plt.legend()
plt.grid(True)
plt.show()
```

```python
n = len(kumpas_degerleri)
sapmalar = kumpas_degerleri - ortalama
sigma = np.sqrt(np.sum(sapmalar**2) / (n * (n - 1)))

print(f"Hesaplanan Ortalama: {ortalama:.4f} cm")
print(f"Hesaplanan Standart Sapma: {sigma:.4f} cm")

plt.plot(olcum_no, kumpas_degerleri, 'o', color='blue', label='Ölçümler')
plt.axhline(ortalama, color='red', linestyle='-', label=f'Ortalama: {ortalama:.3f} cm')

# Hata sınırları (Ortalama ± Standart Sapma)
plt.axhline(ortalama + sigma, color='gray', linestyle=':', label=f'±σ Sınırı: {sigma:.3f} cm')
plt.axhline(ortalama - sigma, color='gray', linestyle=':')
plt.fill_between(olcum_no, ortalama - sigma, ortalama + sigma, color='red', alpha=0.1)

plt.xlabel('Ölçüm Sırası')
plt.ylabel('Uzunluk (cm)')
plt.title(f'Kumpas Ölçüm Analizi: L = ({ortalama:.2f} ± {sigma:.2f}) cm')
plt.legend()
plt.grid(True)
plt.show()
```
