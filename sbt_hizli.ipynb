```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x_olcum = np.array([0.0, 2.4, 4.7, 7.1, 9.3, 11.7, 14.2,16.6,19.0,21.4])
n_nokta = len(x_olcum)
t_olcum = np.arange(n_nokta) * 0.05
egim, kesisim = np.polyfit(t_olcum, x_olcum, 1)
t_fit = np.linspace(t_olcum[0], t_olcum[-1], 100)
x_fit = egim * t_fit + kesisim
plt.figure(figsize=(8, 5))
plt.scatter(
    t_olcum,
    x_olcum,
    color="crimson",
    label="Deneysel Veriler",
    zorder=3,
    marker="o",
)
plt.plot(
    t_fit,
    x_fit,
    color="navy",
    linestyle="--",
    label=f"Fit Doğrusu (v = {egim:.2f} cm/s)",
)

plt.title("Sabit Hızlı Doğrusal Hareket: Konum - Zaman Grafiği", fontsize=12)
plt.xlabel("Zaman, $t$ (s)", fontsize=11)
plt.xticks(np.arange(0.0, 0.50, 0.05))
plt.ylabel("Konum, $x$ (cm)", fontsize=11)
plt.yticks(x_olcum)
plt.grid(True, linestyle=":", alpha=0.7)
plt.legend(fontsize=10)
plt.tight_layout()

plt.show()

print(f"Hesaplanan Hız (Eğim): {egim:.4f} cm/s")
    
```
