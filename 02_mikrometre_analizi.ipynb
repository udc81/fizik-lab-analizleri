```python
import matplotlib.pyplot as plt
import numpy as np

veriler = np.array([
    19.48,
    19.48,
    19.47,
    19.55,
    19.54,
    19.54,
    19.53,
    19.50,
    19.65,
    19.57,
    19.55,
    19.56,
    19.63,
    19.53,
    19.55,
    19.57,
    19.49,
    19.48,
    19.48,
    19.53,
])

print('Veri sayısı:', len(veriler))
print('Ortalama:', np.mean(veriler))
```

```python
degerler, sayilar = np.unique(veriler, return_counts=True)

print("Ölçülen farklı değerler:", degerler)
print("Kaçar tane oldukları   :", sayilar)
```

```python
plt.bar(degerler, sayilar, width=0.005, align='center')
plt.xticks(degerler, rotation=45)
plt.xlabel('Ölçülen Değer (mm)')
plt.ylabel('Kaç Tane Ölçüldü (Adet)')
plt.show()
```
