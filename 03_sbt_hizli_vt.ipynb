```python
import matplotlib.pyplot as plt
import numpy as np

x_olcum = np.array([0.0, 2.4, 4.7, 7.1, 9.3, 11.7, 14.2, 16.6, 19.0, 21.4])
dt = 0.05  
t_olcum = np.arange(len(x_olcum)) * dt
delta_x = np.diff(x_olcum)
v_aralik = delta_x / dt
t_noktalar = t_olcum[1:]
v_fit, _ = np.polyfit(t_olcum, x_olcum, 1)
plt.figure(figsize=(9, 5))
plt.scatter(
    t_noktalar,
    v_aralik,
    color="crimson",
    s=50,
    zorder=3,
    label="Aralık Hızları ($\Delta x / \Delta t$)",
)

# 2. Noktaların üstüne sayıları yazdır(AI)
for t, v in zip(t_noktalar, v_aralik):
    plt.annotate(
        f"{v:.1f}",
        (t, v),
        textcoords="offset points",
        xytext=(0, 7),
        ha="center",
        fontsize=9,
        fontweight="bold",
        color="darkred",
    )
plt.axhline(
    y=v_fit,
    color="navy",
    linestyle="--",
    linewidth=1.5,
    label=f"Fit Doğrusu ($v = {v_fit:.2f}$ cm/s)",
)
plt.title("Hız - Zaman Grafiği", fontsize=12)
plt.xlabel("Zaman, $t$ (s)", fontsize=11)
plt.ylabel("Hız, $v$ (cm/s)", fontsize=11)
plt.xticks(t_olcum)
plt.xlim(-0.02, 0.48)  # Kenarlardan biraz boşluk bırak(AI)
plt.yticks(np.arange(40, 56, 2))
plt.ylim(40, 54)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(loc="upper right")
plt.tight_layout()

plt.show()
```
